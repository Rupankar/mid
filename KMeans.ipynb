# Practical 7: K-means Clustering and Elbow Method

**Aim:** Run K-means for K = 2, 3, 4, 5, 6, plot WCSS, and use the elbow method to choose a possible K.

K-means is unsupervised: the clustering algorithm uses `X` and does not train on the species labels.

## Step 1 — Import libraries

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

## Step 2 — Load the Iris features

```python
data = load_iris()
X = data.data

print("Dataset shape:", X.shape)
```

## Step 3 — Standardize features

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Step 4 — Fit K-means for K = 2 to 6 and record WCSS

```python
k_values = range(2, 7)
wcss = []

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    wcss.append(model.inertia_)
    print(f"K = {k}, WCSS = {model.inertia_:.2f}")
```

## Step 5 — Plot the elbow curve

```python
plt.plot(list(k_values), wcss, marker="o")
plt.xlabel("Number of clusters (K)")
plt.ylabel("WCSS (inertia)")
plt.title("Elbow Method for K-means")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

## Step 6 — Visualize clusters for a chosen K

```python
chosen_k = 3
kmeans = KMeans(n_clusters=chosen_k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled)

plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=cluster_labels)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            marker="X", s=200, label="Centroids")
plt.xlabel("Standardized sepal length")
plt.ylabel("Standardized sepal width")
plt.title(f"K-means clusters (K={chosen_k}; first two features shown)")
plt.legend()
plt.show()
```

## Interpretation
WCSS generally decreases as K increases. Look for an elbow where the reduction begins to slow; the elbow is a heuristic rather than a guaranteed unique answer. The scatter plot shows only two of the four features.
