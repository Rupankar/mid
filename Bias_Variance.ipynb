# Practical 10: Bias–Variance Trade-off

**Aim:** Train Decision Tree regressors with different maximum depths on noisy data, plot training and test errors, and identify a depth with good generalization.

## Step 1 — Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error
```

## Step 2 — Generate a noisy regression dataset

```python
rng = np.random.default_rng(42)
X = np.sort(10 * rng.random((250, 1)), axis=0)
y = np.sin(X).ravel() + rng.normal(0, 0.25, size=len(X))

plt.scatter(X, y, s=15)
plt.xlabel("X")
plt.ylabel("y")
plt.title("Noisy synthetic regression data")
plt.show()
```

## Step 3 — Split into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

## Step 4 — Train regressors at different depths

```python
depths = range(1, 16)
train_errors = []
test_errors = []

for depth in depths:
    model = DecisionTreeRegressor(max_depth=depth, random_state=42)
    model.fit(X_train, y_train)

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_errors.append(mean_squared_error(y_train, train_pred))
    test_errors.append(mean_squared_error(y_test, test_pred))
```

## Step 5 — Plot training and test error against depth

```python
plt.plot(list(depths), train_errors, marker="o", label="Training MSE")
plt.plot(list(depths), test_errors, marker="o", label="Testing MSE")
plt.xlabel("Maximum tree depth")
plt.ylabel("Mean Squared Error")
plt.title("Bias–Variance Trade-off")
plt.xticks(list(depths))
plt.legend()
plt.grid(True)
plt.show()
```

## Step 6 — Find the depth with the lowest test MSE

```python
best_index = int(np.argmin(test_errors))
print("Depth with lowest test MSE:", list(depths)[best_index])
print("Lowest test MSE:", test_errors[best_index])
```

## Interpretation
Shallow trees may underfit (high bias); very deep trees may overfit (high variance). The lowest test MSE is a useful choice for this particular split, not a universal best depth.
