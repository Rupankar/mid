# Practical 8: Principal Component Analysis (PCA)

**Aim:** Reduce the four numerical Iris features to two principal components and visualize species separation.

PCA is an unsupervised dimensionality-reduction technique. Labels are used only to color the visualization.

## Step 1 — Import libraries

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Step 2 — Load the dataset

```python
data = load_iris()
X = data.data
y = data.target
print("Original shape:", X.shape)
```

## Step 3 — Standardize the features

```python
X_scaled = StandardScaler().fit_transform(X)
```

## Step 4 — Reduce four features to two components

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Reduced shape:", X_pca.shape)
print("Explained variance ratio:", pca.explained_variance_ratio_)
print("Total variance explained:", pca.explained_variance_ratio_.sum())
```

## Step 5 — Plot the two principal components

```python
for class_id, species in enumerate(data.target_names):
    plt.scatter(
        X_pca[y == class_id, 0],
        X_pca[y == class_id, 1],
        label=species
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Iris Dataset")
plt.legend()
plt.show()
```

## Interpretation
PCA constructs new axes that capture as much variance as possible. The explained variance ratios indicate how much of the dataset's variance is represented by each component. Species labels are not used when fitting PCA.
