# Practical 9: Independent Component Analysis (ICA) for Signal Separation

**Aim:** Mix two synthetic audio-like signals, apply FastICA, and compare recovered signals with the originals.

This notebook creates synthetic signals rather than loading real audio files.

## Step 1 — Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import FastICA
```

## Step 2 — Create two synthetic source signals

```python
rng = np.random.default_rng(42)
n_samples = 2000
time = np.linspace(0, 8, n_samples)

source_1 = np.sin(2 * time)  # smooth periodic signal
source_2 = np.sign(np.sin(3 * time))  # square-wave-like signal
S = np.c_[source_1, source_2]
S = (S - S.mean(axis=0)) / S.std(axis=0)

plt.figure(figsize=(10, 4))
plt.plot(time, S[:, 0], label="Source 1")
plt.plot(time, S[:, 1], label="Source 2", alpha=0.7)
plt.title("Original synthetic sources")
plt.legend()
plt.show()
```

## Step 3 — Mix the source signals

```python
mixing_matrix = np.array([[1.0, 0.5], [0.4, 1.0]])
X_mixed = S @ mixing_matrix.T

plt.figure(figsize=(10, 4))
plt.plot(time, X_mixed[:, 0], label="Mixture 1")
plt.plot(time, X_mixed[:, 1], label="Mixture 2", alpha=0.7)
plt.title("Observed mixtures")
plt.legend()
plt.show()
```

## Step 4 — Apply FastICA

```python
ica = FastICA(n_components=2, random_state=42, max_iter=1000, tol=0.001)
S_recovered = ica.fit_transform(X_mixed)
```

## Step 5 — Plot recovered sources

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
axes[0].plot(time, S_recovered[:, 0])
axes[0].set_title("Recovered component 1")
axes[1].plot(time, S_recovered[:, 1])
axes[1].set_title("Recovered component 2")
axes[1].set_xlabel("Time")
plt.tight_layout()
plt.show()
```

## Step 6 — Compare source recovery (correlation)

```python
# ICA components may be returned in a different order and with different signs/scales.
for i in range(2):
    correlations = [abs(np.corrcoef(S[:, j], S_recovered[:, i])[0, 1]) for j in range(2)]
    print(f"Recovered component {i + 1}: absolute correlations with sources = {correlations}")
```

## Interpretation
ICA attempts to recover statistically independent source signals from observed mixtures. Components can be returned in a different order and with different signs or scales, so compare using absolute correlation rather than point-by-point equality.
